```python
#sarcasm.json
!gdown --id 1xRU3xY5-tkiPGvlz5xBJ18_pHWSRzI4v

import json
import pandas as pd
data = pd.read_json('sarcasm.json')
data.head()
```

```python
import re

def clean_text(text):
    text = re.sub(r'http\S+|www\S+|https\S+', '', text) # Remove URLs
    text = text.lower() # Convert to lowercase    # remove special characters
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    # remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()
    return text

data['cleaned_headline'] = data['headline'].apply(clean_text)

print("Original Headlines:")
print(data['headline'].head())
print("\nCleaned Headlines:")
print(data['cleaned_headline'].head())
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Initialize a Tokenizer
# Using a reasonable num_words for a larger dataset
# The actual vocabulary size will be len(tokenizer.word_index) + 1
vocab_size_limit = 10000 # Example: limit to top 10000 words
tokenizer = Tokenizer(num_words=vocab_size_limit, oov_token='<unk>')

# Fit the tokenizer on the cleaned headlines
tokenizer.fit_on_texts(data['cleaned_headline'])

# Convert headlines to sequences
sequences = tokenizer.texts_to_sequences(data['cleaned_headline'])

# Determine the maximum sequence length
max_sequence_len = max([len(x) for x in sequences])

# Pad the sequences to a uniform length
padded_sequences = pad_sequences(sequences, maxlen=max_sequence_len, padding='post', truncating='post')

# Get the vocabulary size (actual unique words + 1 for OOV token)
vocabulary_size = len(tokenizer.word_index) + 1

print(f"Vocabulary size: {vocabulary_size}")
print(f"Maximum sequence length: {max_sequence_len}")
print(f"Shape of padded sequences: {padded_sequences.shape}")
print("First 5 original headlines:")
print(data['cleaned_headline'].head().to_list())
print("\nFirst 5 sequences:")
print(sequences[:5])
print("\nFirst 5 padded sequences:")
print(padded_sequences[:5])
```

```python
import numpy as np

# Extract the 'is_sarcastic' column
labels = data['is_sarcastic'].to_numpy()

# Print the first few elements and the shape of the labels array
print("First 5 labels:", labels[:5])
print("Shape of labels array:", labels.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(padded_sequences, labels, test_size=0.2, random_state=42)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GlobalAveragePooling1D, Dense

# 3. Initialize a Sequential model
model = Sequential()

# 4. Add an Embedding layer
# Using vocabulary_size determined from previous steps
model.add(Embedding(input_dim=vocabulary_size, output_dim=32))

# 5. Add a GlobalAveragePooling1D layer
model.add(GlobalAveragePooling1D())

# 6. Add a Dense layer with 1 unit and sigmoid activation
model.add(Dense(1, activation='sigmoid'))

# Explicitly build the model with the expected input shape
# The input shape for a sequence model is (batch_size, sequence_length)
# None for batch_size indicates it can handle any batch size
model.build(input_shape=(None, max_sequence_len))

# 7. Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# 8. Print the model summary
model.summary()
```

```python
history = model.fit(X_train, y_train, epochs=10, validation_data=(X_test, y_test))
print("Model training complete.")
```

```python
loss, accuracy = model.evaluate(X_test, y_test)
print(f"Test Loss: {loss:.4f}")
print(f"Test Accuracy: {accuracy:.4f}")
```

## Build and Compile Model (with Embedding)




```python
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Embedding, GlobalAveragePooling1D, Dense

# vocabulary_size and max_sequence_len are available from previous steps
# vocabulary_size: 28386, max_sequence_len: 39

# 1. Define the Input layer explicitly
inputs_with_embedding = tf.keras.Input(shape=(max_sequence_len,))

# 2. Add an Embedding layer
# Using vocabulary_size determined from previous steps
embedding_layer = Embedding(input_dim=vocabulary_size, output_dim=32)(inputs_with_embedding)

# 3. Add a GlobalAveragePooling1D layer
global_avg_pooling_layer = GlobalAveragePooling1D()(embedding_layer)

# 4. Add a Dense output layer with 1 unit and 'sigmoid' activation for binary classification
outputs_with_embedding = Dense(1, activation='sigmoid')(global_avg_pooling_layer)

# Create the model using the Functional API
model_with_embedding = Model(inputs=inputs_with_embedding, outputs=outputs_with_embedding)

# 5. Compile the model
model_with_embedding.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# 6. Print the model's summary
model_with_embedding.summary()
```

## Train Model (with Embedding)



```python
history_with_embedding = model_with_embedding.fit(X_train, y_train, epochs=10, validation_data=(X_test, y_test))
print("Model with embedding training complete.")
```

```python
import numpy as np
import pandas as pd
import matplotlib
```

a) From Scratch

```python
# Génération de données synthétiques
np.random.seed(42)
X = np.random.uniform(20, 200, 100)  # superficie
true_w0, true_w1 = 5000, 150
y = true_w0 + true_w1 * X + np.random.normal(0, 1000, 100)  # prix avec bruit

# Initialisation des poids
w0, w1 = 0, 0
learning_rate = 6e-5
n_epochs = 5000

# Descente de gradient
for epoch in range(n_epochs):
    y_pred = w0 + w1 * X
    error = y_pred - y
    w0 -= learning_rate * 2 * error.mean()
    w1 -= learning_rate * 2 * (error * X).mean()
    if epoch % 500 == 0:
        print(f"Epoch {epoch}: w0={w0:.2f}, w1={w1:.2f}")

print(f"w0: {w0:.2f}, w1: {w1:.2f}")
```

b) Scikit-Learn

```python
from sklearn.linear_model import LinearRegression

# Reshape X for scikit-learn (expects 2D array)
X_reshaped = X.reshape(-1, 1)

# Create and fit the model
model = LinearRegression()
model.fit(X_reshaped, y)

# Get parameters
sklearn_w0 = model.intercept_
sklearn_w1 = model.coef_[0]
print(f"w0 (intercept): {sklearn_w0:.2f}, w1 (coef): {sklearn_w1:.2f}")
```

c) TensorFlow

```python
import tensorflow as tf

# Créer le modèle linéaire
tf_model = tf.keras.Sequential([
    tf.keras.layers.Dense(1, input_shape=(1,))
])

# Compiler le modèle
tf_model.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=learning_rate), loss='mse')

# Entraîner le modèle
tf_model.fit(X_reshaped, y, epochs=n_epochs, verbose=0)

# Récupérer les poids
tf_w1 = tf_model.layers[0].kernel.numpy()[0][0]
tf_w0 = tf_model.layers[0].bias.numpy()[0]
print(f"w0 (tf): {tf_w0:.2f}, w1 (tf): {tf_w1:.2f}")
```
